```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_excel('Ecommerce_Unclean_Project.xlsx')
```

```python
# Dataset Shape
df.shape
```

```python
# Check Missing values
df.isnull().sum()
```

```python
# Check Duplicate Records
df.duplicated().sum()
```

```python
# Remove Duplicate Records
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
# Replace Invalid values
df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
# Remove extra space
df = df.apply(lambda x:x.str.strip() if x.dtype=='object' else x)
```

```python
# Standarize Text Format
df['Customer_Name'] = df['Customer_Name'].str.title()
df['City'] = df['City'].str.title()
df['State'] = df['State'].str.title()
```

```python
# Validate Email address
df = df[df['Email'].str.contains('@',na=False)]
```

```python
# Convert date column
df['Order_Date'] = pd.to_datetime(
    df['Order_Date'],
    dayfirst=True,
    errors='coerce'
)
```

```python
# Convert numeric columns
cols = ['Qty','Unit_Price','Discount']

for c in cols:
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
# Remove Invalid Quantity
df = df[df['Qty']>0]
```

```python
# Fill missing values
df['Discount'] = df['Discount'].fillna(0)
df['Phone'] = df['Phone'].fillna('Unknown')
df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# Feature Engineering
df['Sales']=df['Qty']*df['Unit_Price']
```

```python
df
```

```python
# Net amount
df['Net_Amount']=df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
df
```

```python
# Profit
df['Profit']=df['Net_Amount']*0.20
```

```python
df
```

```python
# Order month
df['Month']=df['Order_Date'].dt.month_name()
```

```python
df
```

```python
# Order year
df['Year']=df['Order_Date'].dt.year
```

```python
df
```

```python
# weekday
df['Weekday']=df['Order_Date'].dt.day_name()
```

```python
df
```

```python
# Final Data Validation
df.info()
```

```python
# Export clean data
df.to_csv('Clean_Ecommerce.csv',index=False)
```
